CALCULADORA DE TIEMPO EN PLATAFORMAS DIGITALES 

```python


nombre = input("Ingrese su nombre: ")
red1 = "Facebook"
red2 = "Instagram"
red3 = "TikTok"
red4 = "YouTube"
red5 = "Twitter"
print(f"Hola {nombre}, vamos a calcular el porcentaje de tiempo que dedicas a las redes sociales.")

input_red1 = input("Ingrese el tiempo en horas para Facebook: ")
input_red2 = input("Ingrese el tiempo en horas para Instagram: ")
input_red3 = input("Ingrese el tiempo en horas para TikTok: ")
input_red4 = input("Ingrese el tiempo en horas para YouTube: ")
input_red5 = input("Ingrese el tiempo en horas para Twitter: ")
suma_redes = float(input_red1) + float(input_red2) + float(input_red3) + float(input_red4) + float(input_red5)
print(f"El tiempo total dedicado a redes sociales es: {suma_redes} horas.")

float_red1 = float(input_red1)
float_red2 = float(input_red2)
float_red3 = float(input_red3)
float_red4 = float(input_red4)
float_red5 = float(input_red5)

total_redes = float_red1 + float_red2 + float_red3 + float_red4 + float_red5
porcentaje = (total_redes / 24) * 100

print(f"El porcentaje de tiempo dedicado a redes sociales es: {porcentaje:.2f}%")

  
```
