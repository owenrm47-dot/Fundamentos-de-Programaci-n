Extra 1: Control de aforo con break y continue (While)

```python
total_acumulado = 0
boleto = 1


while total_acumulado < 100:
        costo = float(input(f"Costo del boleto {boleto}: "))

        if costo == 0:
            print("Menor de 3 años, sin cargo")
            boleto += 1
            continue

        total_acumulado += costo
        boleto += 1

        if total_acumulado >= 100:
            print("Aforo alcanzado, se detiene el registro")
            break

print(f"Total acumulado: ${total_acumulado:.2f}")

if total_acumulado >= 100:
        print("Aforo alcanzado, se detiene el registro")
boleto += 1

print(f"Total acumulado: ${total_acumulado:.2f}")
```

Extra 2: Estadistica de visitantes con for

```python
contador_adultos = 0 
suma_edades = 0

total_visitantes = int(input("¿Cual es el numero total de visitantes?"))

for i in range(total_visitantes):
    edad = int(input(f"Ingrese la edad del visitante {i + 1}: "))

    suma_edades += edad 
    if edad > 18:
        contador_adultos += 1

if total_visitantes > 0:
    promedio_edad = suma_edades / total_visitantes 
else:
    promedio_edad = 0.0

print(f"Adultos: {contador_adultos:}")
print(f"Promedio de edad de los visitantes: {promedio_edad:.2f}")
```

Extra 4: Piramide de asteriscos con for 

```python
altura = int(input("Altura de la piramide: "))

for i in range(1, altura + 1):
    for j in range (i):
        print("*", end="")
    print()
```
