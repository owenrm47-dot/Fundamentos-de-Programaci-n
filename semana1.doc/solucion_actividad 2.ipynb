```python
personas = int(input("Ingresa el numero de visitantes"))

precio_0_3años = 0
precio_de_17años = 30
precio_mayor_18años = 45
total_general = 0

descuento_Adultomayor = 0.12
descuento_profesor_y_estudiante = 0.10


for m in range(1, personas +1):
    print (f"\nPersonas{m}")
    edad = int(input("¿Cual es la edad de los visitantes?"))

    if edad < 3:
        print("Es menor de 3 años, pasa gratis")
        continue 
    elif edad <= 17:
        precio = precio_de_17años
    else: 
        precio = precio_mayor_18años

    tipo = input("¿Que tipo de visitante eres?")
    if edad >= 60 and tipo == "adulto mayor":
        descuento = precio * descuento_Adultomayor
        porcentaje = 12
    elif tipo == "profesor" or tipo == "estudiante":
        descuento = precio * descuento_profesor_y_estudiante
        porcentaje = 10
    else:
        descuento = 0
        porcentaje = 0
        descuento = precio * descuento_profesor_y_estudiante
        porcentaje = 10 
    
    total = precio - descuento

    print(f"Precio : ${precio:.2f}")
    print(f"Descuento: {porcentaje:.2f}%")
    print(f"Monto del descuento: ${descuento:.2f}")
    print(f"Total a pagar: ${total:.2f}")


    total_general += total

    print(f"Total general ${total_general:.2f}")
```
