Ejercicio 7: Desglose de billetes

```python
monto = int(input("Ingresa el monto (múltiplo de 10): "))

b500 = monto // 500
monto %= 500

b200 = monto // 200
monto %= 200

b100 = monto // 100
monto %= 100

b50 = monto // 50
monto %= 50

b20 = monto // 20
monto %= 20 

b10 = monto // 10
monto %= 10 

print(f"Billetes de $500: {b500}")
print(f"Billetes de $200: {b200}")
print(f"Billetes de $100: {b100}")
print(f"Billetes de $50: {b50}")
print(f"Billetes de $20: {b20}")
print(f"Billetes de $10: {b10}")
```

Ejercicio 8: Potencia sin usar **

```python
base = int(input("Ingresa la base: "))
exponente = int(input("Ingresa el exponente entero positivo: "))

resultado = 1

for _ in range(exponente):
    resultado *= base 
print(f"{base} elevado a {exponente} = {resultado}")
```

Ejercicio 9: El mayor de N números

```python
n = int(input("¿Cuantos numeros se capturan?: "))

mayor = None 

for i in range(n):
    numero = int(input(f"Ingresa el numero {i + 1}: "))
    if mayor is None or numero > mayor:
        mayor = numero 
print(f"El mayor numero visto es: {mayor}")
```

Ejercicio 10: Múltiplos de K con range y paso

```python
k = int(input("Ingresa un numero entero: "))
n = int(input("Ingresa otro numero entero:"))

for i in range(k, n, k):
    print(i, end=" ")
print()
```

Ejercicio 11: Serie alternada


```python
n = int(input("Ingresa un numero entero:"))

suma = 0 

for i in range(1, n + 1):
    if i % 2== 0:
        suma += i
    else: suma -= i 
print(f"El resultado de la serie es: {suma}")
        
```

Ejercicio 12: Serie armónica

```python
n = int(input("Ingresa un numero entero:"))

suma = 0

for i in range(1, n +1):
    suma  += 1 / i 
print(f"La suma de la serie es: {suma:.2f}")
```
