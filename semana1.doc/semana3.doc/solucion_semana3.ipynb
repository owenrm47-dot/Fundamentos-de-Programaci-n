Ejercicio 3: Contador de dígitos

```python
num = int(input("Ingresa un numero entero positvos:"))
original_num = num 
contador = 0

if num == 0:
    contador == 0
else: 
    while num > 0:
        num //= 10
        contador += 1
    num = original_num
print(f"El numero {num} tiene {contador} digitos")
```
