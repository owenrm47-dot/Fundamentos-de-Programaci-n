```python

nombre_estudiante = input("Ingrese el nombre completo del alumno: ")
    
edad_estudiante = int(input("Ingrese la edad del estudiante: "))
    
if edad_estudiante >= 15:
        
        puntaje_examen = int(input("Ingrese el puntaje del examen: "))
        
        if puntaje_examen >= 60:
            print("El alumno", nombre_estudiante, "es asignado al nivel intermedio.")
        else:
            print("El alumno", nombre_estudiante, "es asignado al nivel basico.")
def gestionar_inscripcion():
    print(f"Inscripción de {nombre_estudiante} completada.")

gestionar_inscripcion()
```
