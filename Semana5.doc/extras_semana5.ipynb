Extra 1: Sistema de calificaciones con tuplas


```python
calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

print(f"Tercera Calificacion: {calificaciones[-3]}")

extra1 = float(input("Ingresa el primer numero extra:"))
extra2 = float(input("Ingresa el segundo numero extra:"))

def suma_tupla(t):
    total = 0 
    for n in t:
        total += n 
    return total

calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0, extra1, extra2)
print(f"Nueva tupla: {calificaciones}")

calificaciones_extendida = calificaciones + (extra1, extra2)
lista = list(calificaciones_extendida)
lista.sort(reverse=True)
print(f"Lista ordenada: {lista}")

resultado = suma_tupla(calificaciones)
print(f"Suma total: {resultado}")
```

Extra 2: Agenda de contactos con búsqueda

```python
agenda = {
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"
}

agregar_contacto = str(input("Ingresa el nombre del contacto a agregar: ")).strip()
agregar_numero = int(input(f"Ingresa el número de teléfono para {agregar_contacto}: ")).strip()
agenda[agregar_contacto] = agregar_numero

print(f"Contactos registrados: {', '.join(agenda.keys())}") 
buscar_contacto = input("Ingresa el nombre del contacto a buscar: ").strip()
telefono = agenda.get(buscar_contacto)

if telefono is not None:
    print(f"El teléfono de {buscar_contacto} es: {telefono}")
```

Desafío Extra 1: Lista de espera con tuplas

```python

espera = ("Maria", "Jose", "Carlos", "Lucia", "Pedro")
print(espera[-3])

nuevo1 = str(input("Ingresa el primer nombre: "))
nuevo2 = str(input("Ingresa el segundo nombre: "))

nuevos_nombres = (nuevo1, nuevo2)
espera_combinada = espera + nuevos_nombres


lista_ordenada = list(espera_combinada)
lista_ordenada.sort()

print(f"Lista ordenada alfabeticamente: {lista_ordenada}")
print(f"Cantidad total de personas: {len(lista_ordenada)}")
```

Desafío Extra 2: Contador de votos con diccionario


```python
votos ={"Rojo": 0, "Azul": 0, "Verde": 0}

for _ in range(5):
    color = input("Ingresa el color de tu voto (Rojo/azul/Verde): ")
    if color in votos:
          votos[color] += 1
print("\nResultados finales:")
for color, total in votos.items():
    print(f"-{color}: {total} votos")
```
